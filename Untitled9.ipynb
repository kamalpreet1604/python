```python
import pandas as pd
#import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression #using liaberery
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor,GradientBoostingRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import train_test_split #using this for spliting data
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error#
import numpy as np
import warnings
warnings.filterwarnings('ignore')
df=pd.read_csv("house_data_large.csv") #houseing data


x=df[["Sqft","Bedrooms","Bathrooms","YearBuilt","LocationCode",]]#factors provide
y=df[["Price"]]

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
model={"Linear regression":LinearRegression(),"Decision Tree Regressor":DecisionTreeRegressor(),
       "Random Forest Regressor":RandomForestRegressor(),",Gradient Boosting Regressor":GradientBoostingRegressor(),
       "K Neighbors Regressor":KNeighborsRegressor()
}


user_input=np.array([[3774,2,4,2006,1]])

for name, model in model.items():
    model.fit(x_train, y_train)
    pred = model.predict(x_test)
    user_pred = model.predict(user_input)

    print(name)
    print("============================================================")
    print("Predicted Price:", user_pred)
    print("mae:", mean_absolute_error(y_test, pred))
    print("mse:", mean_squared_error(y_test, pred))
    print("rmse:", mean_squared_error(y_test, pred))
```
