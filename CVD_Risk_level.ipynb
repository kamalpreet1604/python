```python
# finding categorial columns in dataset
categorical_cols = df.select_dtypes(include="object").columns
print("Categorical Columns:")
print(categorical_cols)
print(df[categorical_cols].isnull().sum())
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
#labelEncoder as we have three values
df["CVD Risk Level"] = le.fit_transform(df["CVD Risk Level"])
# pandas dummies when we have 2 values
cols_to_encode = [
  "Sex",
    "Smoking Status",
    "Diabetes Status",
    "Physical Activity Level",
    "Family History of CVD",
    "Blood Pressure Category"
]
df_encoded = pd.get_dummies(df, columns=cols_to_encode, drop_first=True)
# print(df_encoded.head())
# print(df_encoded.shape)
```

```python

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
import warnings
warnings.filterwarnings("ignore")

# Drop target columns and the problematic 'Blood Pressure (mmHg)' column
x = df_encoded.drop(["CVD Risk Level", "CVD Risk Score", "Blood Pressure (mmHg)"], axis=1)
y = df_encoded["CVD Risk Level"]

# Train-Test Split (should be before scaling)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

#scaling
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

# Example user input (encoded format)
# (Make sure these columns exist in x.columns)
user_data = {
    "Age": 32,
    "Weight (kg)": 69.1,
    "Height (m)": 1.71,
    "BMI": 23.6,
    "Abdominal Circumference (cm)": 86.2,
    "Total Cholesterol (mg/dL)": 248,
    "HDL (mg/dL)": 78,
    "Fasting Blood Sugar (mg/dL)": 111,
    "Waist-to-Height Ratio": 0.50,
    "Systolic BP": 125,
    "Diastolic BP": 79,
    "Estimated LDL (mg/dL)": 140,
    "Sex_M": 0,
    "Smoking Status_Y": 0,
    "Diabetes Status_Y": 0,
    "Physical Activity Level_Low": 1,
    "Physical Activity Level_Moderate": 0,
    "Family History of CVD_Y": 0,
    "Blood Pressure Category_Hypertension Stage 1": 0,
    "Blood Pressure Category_Hypertension Stage 2": 0,
    "Blood Pressure Category_Normal": 0
}

user_df = pd.DataFrame([user_data])
user_df = user_df.reindex(columns=x.columns, fill_value=0)
user_scaled = scaler.transform(user_df)

# Model training
# user_df = pd.DataFrame([user_data]) # This line is redundant, user_df already created above
# user_df = user_df.reindex(columns=x.columns, fill_value=0) # This line is redundant

#Train Logistic Regression (ON SCALED DATA)
model = LogisticRegression(max_iter=2000)
model.fit(x_train_scaled, y_train)

#Predict (ONLY ON SCALED DATA)
y_pred = model.predict(x_test_scaled)
user_pred = model.predict(user_scaled)

# Define risk_map based on LabelEncoder's classes if available, otherwise assume a common mapping
# Assuming 'High', 'Low', 'Medium' are encoded as 0, 1, 2 respectively by default LabelEncoder
# For accurate mapping, you might want to inspect 'le.classes_' if 'le' is accessible
risk_map = {0: 'High', 1: 'Low', 2: 'Medium'} # Adjust this based on actual LabelEncoder mapping
print("Accuracy:", accuracy_score(y_test, y_pred))
print("User CVD Risk Level:", risk_map[user_pred[0]])
print("Model: Logistic Regression")
print("User CVD Risk Level:", user_pred[0])
```

```python
#Accuracy
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
# Accuracy
acc = accuracy_score(y_test, y_pred)
print("Accuracy:", acc)
# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:\n", cm)
# Classification Report
print(classification_report(y_test, y_pred))


```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
dt_model = DecisionTreeClassifier(random_state=42, min_samples_split=10)    # prevents overfitting

dt_model.fit(x_train, y_train)
y_pred_dt = dt_model.predict(x_test)

print("Decision Tree Accuracy:", accuracy_score(y_test, y_pred_dt))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_dt))
print(classification_report(y_test, y_pred_dt))
```

```python
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    max_depth=10,
    class_weight="balanced"   # VERY IMPORTANT for health data
)

rf_model.fit(x_train, y_train)
y_pred_rf = rf_model.predict(x_test)
print("Random Forest Accuracy:", accuracy_score(y_test, y_pred_rf))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_rf))
print(classification_report(y_test, y_pred_rf))
gb_model = GradientBoostingClassifier(n_estimators=100, random_state=42)
gb_model.fit(x_train, y_train)
y_pred_gb = gb_model.predict(x_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
knn_model = KNeighborsClassifier(
    n_neighbors=7,     # better than default 5
    weights="distance" # closer neighbors matter more
)
knn_model.fit(x_train_scaled, y_train)
y_pred_knn = knn_model.predict(x_test_scaled)
print("KNN Accuracy:", accuracy_score(y_test, y_pred_knn))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_knn))
print(classification_report(y_test, y_pred_knn))
```
